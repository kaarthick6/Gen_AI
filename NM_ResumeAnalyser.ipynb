```python
import os
import gradio as gr
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage

# Set up OpenRouter API key
os.environ["OPENAI_API_KEY"] = "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"
os.environ["OPENAI_API_BASE"] = "https://openrouter.ai/api/v1"

# Initialize LangChain Chat Model
llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.3)

def analyze_resume(resume_text):
    """Analyzes a resume provided as raw text and gives feedback."""
    if not resume_text.strip():
        return "❌ Please enter resume text for analysis."

    # AI prompt for resume analysis
    system_prompt = (
       "You are an experienced hiring manager and career consultant. "
        "Analyze the given resume thoroughly. Identify and critique specific aspects, including:"
        "\n1️⃣ **Strengths**: What stands out? Provide examples from the resume."
        "\n2️⃣ **Weaknesses**: Where can it improve? Point out missing details."
        "\n3️⃣ **Formatting Issues**: Any structure, clarity, or readability problems?"
        "\n4️⃣ **Skill Gaps**: What crucial skills or certifications are missing for career growth?"
        "\n5️⃣ **Suggested Rewrites**: Rewrite weak bullet points or sections to be more impactful."
        "\n\nBe specific and use examples directly from the resume text."
    )

    # AI Analysis
    response = llm.invoke([
        SystemMessage(content=system_prompt),
        HumanMessage(content=f"Analyze this resume:\n\n{resume_text}")
    ])

    return response.content  # Return AI-generated feedback

# Gradio UI
iface = gr.Interface(
    fn=analyze_resume,
    inputs=gr.Textbox(label="Paste Resume Text Here", lines=15, placeholder="Enter resume content..."),
    outputs=gr.Textbox(label="AI Resume Feedback"),
    title="📄 AI Resume Analyzer",
    description="Paste a resume in text format, and AI will analyze its strengths, weaknesses, and areas for improvement.",
)

# Launch Gradio UI
iface.launch(share=True)
```

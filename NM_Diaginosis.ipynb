```python
import os
import gradio as gr
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage

# Set up OpenRouter API key
os.environ["OPENAI_API_KEY"] = "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"
os.environ["OPENAI_API_BASE"] = "https://openrouter.ai/api/v1"

# Initialize LangChain Chat Model
llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.2)

def diagnose_medical_report(report_text):
    """Analyzes a medical report and provides a possible diagnosis with insights."""
    if not report_text.strip():
        return "❌ Please enter the medical report text for analysis."

    # AI prompt for medical diagnosis
    system_prompt = (
        "You are an expert medical doctor with years of experience in diagnosing diseases "
        "based on medical reports. Given a patient's medical report, provide the following:"
        "\n1️⃣ **Possible Diagnosis** - Likely conditions based on the symptoms & report findings."
        "\n2️⃣ **Key Observations** - Noteworthy details from the report."
        "\n3️⃣ **Suggested Next Steps** - Recommended actions (e.g., further tests, treatments)."
        "\n\n⚠️ Disclaimer: This is an AI-generated analysis and should not replace professional medical advice."
    )

    # AI Analysis
    response = llm.invoke([
        SystemMessage(content=system_prompt),
        HumanMessage(content=f"Analyze this medical report:\n\n{report_text}")
    ])

    return response.content  # Return AI-generated diagnosis & insights

# Gradio UI
iface = gr.Interface(
    fn=diagnose_medical_report,
    inputs=gr.Textbox(label="Paste Medical Report Here", lines=15, placeholder="Enter medical report..."),
    outputs=gr.Textbox(label="AI Diagnosis & Insights"),
    title="🩺 AI Medical Diagnosis Assistant",
    description="Paste a medical report, and AI will analyze it to provide a possible diagnosis and insights. 🚨 **Note:** AI is not a substitute for a licensed medical professional.",
)

# Launch Gradio UI
iface.launch(share=True)
```

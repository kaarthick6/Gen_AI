```python
# Step 1: Install requests if needed (usually available in Colab)
!pip install -q requests

# Step 2: Import libraries
import requests
import json

# Step 3: Set your API key (store it securely in real applications)
OPENROUTER_API_KEY = "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"  

# Step 4: Define the function to call GPT-4 via OpenRouter
def generate_story(prompt, temperature=0.8, max_tokens=500):
    url = "https://openrouter.ai/api/v1/chat/completions"
    headers = {
        "Authorization": f"Bearer {OPENROUTER_API_KEY}",
        "Content-Type": "application/json"
    }
    payload = {
        "model": "openai/gpt-3.5-turbo",  # You can also try 'mistralai/mixtral-8x7b' etc.
        "messages": [
            {"role": "system", "content": "You are a creative storytelling AI. Create engaging, imaginative, and vivid stories."},
            {"role": "user", "content": prompt}
        ],
        "temperature": temperature,
        "max_tokens": max_tokens
    }

    response = requests.post(url, headers=headers, data=json.dumps(payload))

    try:
        result = response.json()
    except Exception as e:
        print("Failed to parse JSON:", e)
        print("Raw response:", response.text)
        return None

    # Debug if 'choices' is missing
    if 'choices' not in result:
        print("API Error or unexpected response:")
        print(json.dumps(result, indent=2))
        return None

    return result['choices'][0]['message']['content']


# Step 5: Example usage
prompt = "Tell me a fantasy story about a swordsman travelling to a village filled with orcs."
story = generate_story(prompt)
print(story)
```
